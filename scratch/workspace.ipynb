```python
# Test the function
batch_size = 2
seq_len = 4
num_heads = 3
head_dim = 8  # Must be even

# Create random input tensor (queries or keys)
x = torch.randn(batch_size, seq_len, num_heads, head_dim)

```

```python
cos = 0.1
sin = 0.4
assert x.ndim == 4  # multihead attention
d = x.shape[3] // 2
x1, x2 = x[..., :d], x[..., d:]
y1 = x1 * cos + x2 * sin
y2 = x1 * (-sin) + x2 * cos
output = torch.cat([y1, y2], 3)
output
```
